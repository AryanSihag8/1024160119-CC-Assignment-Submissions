Q1.  Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.).

1. Convert text to lowercase and remove punctua on using re.

2. Tokenize the text into words and sentences.

3. Split using split() and word_tokenize() and compare how Python split and NLTK’s word_tokenize() differ.

4. Remove stopwords (using NLTK's stopwords list).

5. Display word frequency distribu on (excluding stopwords).

```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from collections import Counter

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Paragraph about boxing
text = """
Boxing is an exciting combat sport that requires strength,
speed, discipline, and strategy. Boxers train every day to
improve their punching power and defensive skills. A good
boxer must maintain fitness, focus, and confidence.
Boxing teaches patience, courage, and respect for opponents.
"""

# 1. Convert to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("Cleaned Text:\n", clean_text)

# 2. Tokenize into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:", words)
print("\nSentence Tokens:", sentences)

# 3. Compare split() and word_tokenize()
print("\nUsing split():", clean_text.split())
print("\nUsing word_tokenize():", word_tokenize(text.lower()))

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in words
    if word not in stop_words and word.isalpha()
]
print("\nWords after Stopword Removal:", filtered_words)

# 5. Display word frequency distribution
frequency = Counter(filtered_words)
print("\nWord Frequency Distribution:")
for word, count in frequency.most_common():
    print(word, ":", count)
```

Q2. Using the same paragraph from Q1:

1. Extract all words with only alphabets using re.findall()

2. Remove stop words using NLTK’s stopword list

3. Perform stemming with PorterStemmer

4. Perform lemma za on with WordNetLemma zer

5. Compare the stemmed and lemma zed outputs and explain when you’d prefer one over the other.

```python
import re
import nltk
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Same boxing paragraph from Q1
text = """
Boxing is an exciting combat sport that requires strength,
speed, discipline, and strategy. Boxers train every day to
improve their punching power and defensive skills. A good
boxer must maintain fitness, focus, and confidence.
Boxing teaches patience, courage, and respect for opponents.
"""

# 1. Extract alphabetic words
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print("Alphabetic Words:\n", words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [w for w in words if w not in stop_words]
print("\nAfter Stopword Removal:\n", filtered_words)

# 3. Perform stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(w) for w in filtered_words]
print("\nStemmed Words:\n", stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(w, pos=wordnet.NOUN)
    for w in filtered_words
]
print("\nLemmatized Words:\n", lemmatized_words)

# 5. Compare outputs
print("\nComparison:")
for original, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:15} Stem: {stem:12} Lemma: {lemma}")
```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).

1. Use CountVectorizer to generate the Bag of Words representa on.

2. Use TfidfVectorizer to compute TF-IDF scores.

3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Three short texts about boxing
texts = [
    "Boxing requires strength speed and discipline",
    "Boxers practice punching defense and fitness",
    "Boxing matches require strategy courage and stamina"
]

# 1. Bag of Words
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

# 2. Calculate TF-IDF scores
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

features = tfidf_vectorizer.get_feature_names_out()
scores = tfidf_matrix.toarray()

print("\nTF-IDF Matrix:")
print(scores)

# 3. Print top 3 keywords from each text
print("\nTop 3 TF-IDF Keywords:")
for i, row in enumerate(scores):
    top_indices = row.argsort()[::-1][:3]

    print(f"\nText {i + 1}: {texts[i]}")
    for index in top_indices:
        print(f"{features[index]}: {row[index]:.4f}")
```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).

1. Preprocess and tokenize both texts.

2. Calculate:

  a. Jaccard Similarity using sets

  b. Cosine Similarity using TfidfVectorizer + cosine_similarity()

  c. Analyze which similarity metric gives be er insights in your case.

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Two short texts
boxing_text = """
Boxing is a combat sport that requires strength and speed.
Boxers practice punches, defense, and fitness every day.
Training improves stamina, discipline, and confidence.
Successful boxers use strategy to win their matches.
"""

football_text = """
Football is a team sport that requires strength and speed.
Players practice passing, defending, and fitness every day.
Training improves stamina, teamwork, and confidence.
Successful players use strategy to win their matches.
"""

# 1. Preprocess and tokenize
def preprocess(text):
    text = text.lower()
    return re.findall(r'\b[a-z]+\b', text)

tokens1 = preprocess(boxing_text)
tokens2 = preprocess(football_text)

print("Boxing Tokens:", tokens1)
print("\nFootball Tokens:", tokens2)

# 2a. Jaccard similarity using sets
set1 = set(tokens1)
set2 = set(tokens2)

jaccard = len(set1 & set2) / len(set1 | set2)

print("\nJaccard Similarity:", round(jaccard, 4))

# 2b. Cosine similarity using TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(
    [boxing_text, football_text]
)

cosine = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine, 4))

# 3. Compare the metrics
print("\nAnalysis:")
print("Jaccard measures the overlap between unique words.")
print("Cosine measures similarity using TF-IDF word weights.")

if cosine > jaccard:
    print("Cosine similarity is higher for these texts.")
elif cosine < jaccard:
    print("Jaccard similarity is higher for these texts.")
else:
    print("Both similarity scores are equal.")
```

Q5. Write a short review for a product or service.

1. Use TextBlob or VADER to find polarity & subjec vity for each review.

2. Classify reviews into Posi ve / Nega ve / Neutral.

3. Create a word cloud using the wordcloud library for all posi ve reviews.

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Download VADER lexicon
nltk.download('vader_lexicon')

# Boxing reviews
reviews = [
    "Boxing training is exciting and improves my fitness.",
    "The boxing coach is excellent and very supportive.",
    "The boxing class was terrible and exhausting.",
    "I disliked the poor training facilities.",
    "The boxing session was okay.",
    "The training session was average."
]

# Initialize sentiment analyzer
sia = SentimentIntensityAnalyzer()

positive_reviews = []

# 1 and 2. Calculate sentiment and classify reviews
for review in reviews:
    scores = sia.polarity_scores(review)
    compound = scores['compound']

    if compound >= 0.05:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif compound <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("\nReview:", review)
    print("Positive score:", scores['pos'])
    print("Negative score:", scores['neg'])
    print("Neutral score:", scores['neu'])
    print("Compound polarity:", scores['compound'])
    print("Classification:", sentiment)

# 3. Generate word cloud from positive reviews
if positive_reviews:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white",
        colormap="Reds"
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Positive Boxing Reviews")
    plt.show()
else:
    print("No positive reviews found.")
```

Q6. Choose your own paragraph (~100 words) as training data.

1. Tokenize text using Tokenizer() from keras.preprocessing.text

2. Create input sequences and build a simple LSTM or Dense model

3. Train the model and generate 2–3 new lines of text star ng from any seed word you provide.

```python
import numpy as np
import nltk
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Training paragraph of approximately 100 words
text = """
Boxing is a demanding sport that combines strength, speed,
skill, and intelligence. Boxers train regularly to improve
their stamina, footwork, balance, and punching technique.
A successful boxer must learn both attack and defense.
Coaches teach athletes to remain focused during difficult
rounds. Practice helps boxers develop confidence and discipline.
Good nutrition and proper rest support recovery after training.
Boxing matches require patience, timing, and careful strategy.
Athletes must respect their opponents and follow the rules.
Consistent training helps a boxer become stronger and more skilled.
Every boxing champion begins with dedication and hard work.
"""

# 2. Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1
sequences = []

# Create input sequences
for line in text.lower().split("\n"):
    line = line.strip()
    if not line:
        continue

    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        sequences.append(token_list[:i + 1])

# Pad sequences
max_sequence_len = max(len(seq) for seq in sequences)
sequences = pad_sequences(
    sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

# Split inputs and target words
X = sequences[:, :-1]
y = sequences[:, -1]

# 3. Build a simple LSTM model
model = Sequential([
    Embedding(total_words, 32, input_length=max_sequence_len - 1),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()

# Train the model
model.fit(X, y, epochs=200, verbose=0)

print("\nModel training completed!")

# 4. Generate text from a seed phrase
def generate_text(seed_text, next_words=10):
    result = seed_text.lower()

    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([result])[0]
        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding="pre"
        )

        predicted = model.predict(token_list, verbose=0)
        predicted_id = int(np.argmax(predicted, axis=1)[0])

        if predicted_id == 0:
            break

        output_word = next(
            (word for word, index in tokenizer.word_index.items()
             if index == predicted_id),
            ""
        )

        if not output_word:
            break

        result += " " + output_word

    return result

# Seed word
seed = "boxing"

# Generate 2–3 lines of text
for i in range(3):
    print(generate_text(seed, next_words=10))
```
