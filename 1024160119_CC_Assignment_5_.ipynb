Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:

a) Addition of 2 in all the element

b) Multiply 3 with all the elements

c) Divide every element by 2

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])

print("Original Array : ", arr1)
print("Addition of 2 in all the element : ", arr1 + 2)
print("Multiply 3 with all the elements  : ", arr1 * 3)
print("Divide every element by 2 : ", arr1 / 2)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

value = np.bincount(x).argmax()
print("x - Most frequent : ", value)
print("Indices : ", np.where(x == value)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

value1 = np.bincount(y).argmax()
print("y - Most frequent : ", value1)
print("Indices : ", np.where(y == value1)[0])
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
Aryan = np.linspace(10, 100, 25)

print("Array : ", Aryan)
print("Dimensions : ", Aryan.ndim)
print("Shape : ", Aryan.shape)
print("Total elements : ", Aryan.size)
print("Data type : ", Aryan.dtype)
print("Total bytes : ", Aryan.nbytes)

print("Transpose using reshape() : ", Aryan.reshape(25, 1))

print("Transpose using T : ", Aryan.reshape(1, 25).T)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_Aryan = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array : ", ucs420_Aryan)

print("Mean : ", np.mean(ucs420_Aryan))
print("Median : ", np.median(ucs420_Aryan))
print("Maximum : ", np.max(ucs420_Aryan))
print("Minimum : ", np.min(ucs420_Aryan))
print("Unique Elements : ", np.unique(ucs420_Aryan))

reshaped_ucs420_Aryan = np.reshape(ucs420_Aryan, (4, 3))

print("\nReshaped Array : ")
print(reshaped_ucs420_Aryan)

resized_ucs420_Aryan = np.resize(ucs420_Aryan, (2, 3))

print("\nResized Array : ")
print(resized_ucs420_Aryan)
```
